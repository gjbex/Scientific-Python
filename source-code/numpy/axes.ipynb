# Aggregation functions

## Aggregation over 2D array

Create a $4 \times 6$ array.

```python
A = np.arange(24).reshape(4, -1)
A
```

Sum over the elements of the columns.

```python
A.sum(axis=0)
```

Sum over the elements of the rows.

```python
A.sum(axis=1)
```

Sum over all elements of the array.

```python
A.sum()
```

## Aggregation over 3D array

Create a $2 \times 3 \times 4$ array.

```python
B = A.reshape(2, 3, -1)
B
```

Sum over axis 0, the result is a $3 \times 4$ array.

```python
B.sum(axis=0)
```

Sum over axis 0 and 1, the result is an array with 4 elements.

```python
B.sum(axis=(0, 1))
```

Sum over axis 0 and 2, the result is an array with 3 elements.

```python
B.sum(axis=(0, 2))
```

The axis apply to other aggregation functions such as `np.min`, `np.max`, `np.prod` and so on.

# Transpose

For multi-dimensional arrays, the `np.transpose` function can transpose over arbitrary axes.

```python
C = np.arange(24).reshape(2, 3, -1)
```

```python
C
```

By default, the first and last axis are swapped.

```python
np.transpose(C)
```

```python
assert(np.allclose(
    C.transpose(),
    np.einsum('ijk -> kji', C)
))
```

However, you can swap any axes by specifying a tuple.

```python
np.transpose(C, axes=(1, 2, 0))
```

```python
assert(np.allclose(
    np.transpose(C, axes=(1, 0, 2)),
    np.einsum('ijk -> jik', C)
))
```

Note that no data is moved during a transpose.

```python
assert(
    C.base is np.transpose(C, axes=(2, 0, 1)).base
)
```

If you want to swap two axes, you can do so using `np.swapaxes`.

```python
assert(np.allclose(
    np.transpose(C, axes=(1, 0, 2)),
    np.swapaxes(C, 0, 1)
))
```
