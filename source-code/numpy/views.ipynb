# Views

Create a 1-dimensional array.

```python
a = np.arange(10)
```

Create a slice.

```python
b = a[2:6]
```

`b` is a view on `a`, so changing an element of `b` will change the corresponding value in `a`.

```python
b[1] = -1
```

```python
b
```

```python
a
```

You can check whether `a` and `b` share memory.

```python
np.shares_memory(a, b)
```

Another way to check is to inspect the `base` attribute.

```python
print(f'{a.base=}, {b.base=}')
```

As `a` is not derived from an array, its `base` is `None`.  Since `b` is a view of `a`, its `base` attribute are the values of `a`.

Another way of checking is using `flags.own`

```python
print(f'{a.flags.owndata=}, {b.flags.owndata=}')
```

# Pitfall

Create a 2D-array using `reshape`.

```python
a = np.arange(24).reshape(4, -1)
```

```python
print(f'{a.base=}')
```

This might come as a surprise at first sight. However, breaking this down into two steps illustrates what happens.

```python
b = np.arange(24)
c = b.reshape(4, -1)
```

```python
print(f'{b.base=}, {c.base=}')
```

Since you use `reshape`, `c` is a view on `b`.  When combining array creation and `reshape`, the result is a view, not an array.

# Fancy indexing

When you use fancy indexing, you can get a new array, rather than a view.

```python
a = np.arange(20)
```

```python
b = a[a % 3 == 0]
```

```python
print(f'{a.base=}, {b.base=}')
```

```python
np.shares_memory(a, b)
```

In case of doubt, you can check using `np.shares_memory`.
