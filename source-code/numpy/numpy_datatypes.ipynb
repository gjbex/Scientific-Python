# numpy data types

numpy supports several floating point types, single precision `np.float32`, double precision `np.float64`, but also half precision `np.float16` and extended precision `np.float128`.

Note that `np.float128` is not a portable format.  Its implementation differs across hardware platforms.

```python
import numpy as np
```

Create a random number genrator.

```python
rng = np.random.default_rng(seed=1234)
```

```python
size = (2_000, 2_000)
```

```python
for type in [np.float16, np.float32, np.float64, np.float128]:
    data = rng.uniform(size=size).astype(type)
    print(f'type = {type}')
    get_ipython().run_line_magic('timeit', 'data.sum()')
```

The differences in running time are as you might expect.

The properties of the various floating point types can be queried using the `np.finfo` function.

```python
for ftype in [np.float16, np.float32, np.float64, np.float128]:
    finfo = np.finfo(ftype)
    print(f'{ftype.__name__}: min. exp. = {finfo.minexp}, '
          f'max. exp. = {finfo.maxexp}, digits = {finfo.precision}')
```
