# numpy data types

NumPy supports several floating-point types, including half precision `np.float16`, single precision `np.float32`, and double precision `np.float64`. Some platforms also provide `np.float128`.

`np.float128` is not portable: it may be absent, and its name does not guarantee IEEE 754 binary128 precision. When it exists, it is an alias for `np.longdouble`, whose actual precision depends on the compiler and platform. The code below therefore adds `np.float128` only when it is available, and uses `np.finfo` to inspect the precision actually provided.

```python
import numpy as np

floating_types = [np.float16, np.float32, np.float64]
try:
    floating_types.append(np.float128)
except AttributeError:
    print("np.float128 is not available on this platform")
```

Create a random number genrator.

```python
rng = np.random.default_rng(seed=1234)
```

```python
size = (2_000, 2_000)
```

```python
for dtype in floating_types:
    data = rng.uniform(size=size).astype(dtype)
    print(f'type = {dtype}')
    get_ipython().run_line_magic('timeit', 'data.sum()')
```

The differences in running time are as you might expect.

The properties of the various floating point types can be queried using the `np.finfo` function.

```python
for dtype in floating_types:
    finfo = np.finfo(dtype)
    print(f'{dtype.__name__}: min. exp. = {finfo.minexp}, '
          f'max. exp. = {finfo.maxexp}, digits = {finfo.precision}')
```
