# Requirements

```python
import numpy as np
```

# Aggregation over 2D array

Create a $4 \times 6$ array.

```python
A = np.arange(24).reshape(4, -1)
A
```

Sum over the elements of the columns.

```python
A.sum(axis=0)
```

Sum over the elements of the rows.

```python
A.sum(axis=1)
```

Sum over all elements of the array.

```python
A.sum()
```

# Aggregation over 3D array

Create a $2 \times 3 \times 4$ array.

```python
B = A.reshape(2, 3, -1)
B
```

Sum over axis 0, the result is a $3 \times 4$ array.

```python
B.sum(axis=0)
```

Sum over axis 0 and 1, the result is an array with 4 elements.

```python
B.sum(axis=(0, 1))
```

Sum over axis 0 and 2, the result is an array with 3 elements.

```python
B.sum(axis=(0, 2))
```

The axis apply to other aggregation functions such as `np.min`, `np.max`, `np.prod` and so on.
