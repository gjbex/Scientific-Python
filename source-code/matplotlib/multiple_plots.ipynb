The goal is to illustrate how to create a plot that consists of subplots, side-by-side, and with multiple rows.

# Requirements

```python
import matplotlib.pyplot as plt
import numpy as np
```

# Data

You want to plot two functions for three values of the damping factor $\mu$.
$$
    \begin{aligned}
        f_1(t, \mu) & = \cos(2\pi t)\exp(-\mu t) \\
        f_2(t, \mu) & = \cos(2\pi t)\exp(-\mu t) + \cos((2\pi + \delta) t)\exp(-\mu t)
    \end{aligned}
$$
The time interval is $[0, 20\pi]$ and $\mu \in \{0.0, 0.1, 0.3\}$ while $\delta = 0.7$.

```python
t = np.linspace(0.0, 20.0*np.pi, 1001)
μ_values = np.array([0.0, 0.1, 0.3])
Y = np.stack((
    np.cos(2.0*np.pi*t)*np.exp(-μ_values[:, None]*t),
    np.cos(2.0*np.pi*t)*np.exp(-μ_values[:, None]*t) + 0.1*np.cos((2.0*np.pi + 0.7)*t)*np.exp(-μ_values[:, None]*t),
))
```

# Plots

Create a plot with 2 rows and 3 columns, setting a figure size, and controlling the layout.  Next, iterate over the 2D index of `axes`, and plot the corresponding data on the axes.

```python
figure, axes = plt.subplots(
    nrows=2, ncols=3,
    figsize=(10, 3),
    layout='constrained'
)
for idx in np.ndindex(axes.shape):
    axes[idx].plot(t, Y[idx])
```

Save the figure to a file.

```python
figure.savefig('multiple_plots.png')
```
